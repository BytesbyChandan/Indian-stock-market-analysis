```python
import pandas as pd
import pymysql as sql
import matplotlib.pyplot as plt
import seaborn as sns
conn = sql.connect(host='localhost', user='root', password='1234', database='stock_analysis')
```

```python
query = '''
        Select ticker, round(max(close),2) as Highest_closing_price
from stock_market
group by ticker;

'''
# Highest closing price for each stock ticker
```

```python
df = pd.read_sql(query, conn)
df
```

```python
# iwant to make it colorful and attractive and also want to add the highest closing price for each stock ticker in a bar chart using matplotlib and seaborn

plt.figure(figsize=(14, 6), facecolor='lightgray')


sns.barplot(x='ticker',
            y='Highest_closing_price', 
            data=df, 
            palette=sns.color_palette('viridis', len(df)), 
            edgecolor='black',
            saturation=0.8,
            orient = 'v',
            gap = 0.2)


plt.title('Highest Closing Price for Each Stock Ticker', color='#330747', fontsize=20)
plt.xlabel('Stock Ticker', color='black', fontsize=16)
plt.ylabel('Highest Closing Price', color='black', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
query = '''Select ticker, sector, round(avg(daily_return_pct),2) as avg_return,
rank() over(partition by sector order by avg(daily_return_pct)) as rnk
from stock_market
group by ticker, sector;'''
```

```python
df = pd.read_sql(query, conn)
df
```

```python
heatmap_data = df.pivot_table(
    index="sector",
    columns="ticker",
    values="avg_return"
)

plt.figure(figsize=(18, 10))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".2f",
    cmap="RdYlGn",
    center=0,
    linewidths=0.5
)

plt.title("Sector-wise Average Daily Stock Returns", fontsize=20)
plt.xlabel("Stock Ticker", fontsize = 14)
plt.ylabel("Sector", fontsize = 14)
plt.tight_layout()
plt.show()
```

```python
query =''' create view stock_performance as
Select ticker, company, sector,
round(avg(close), 2) AS avg_close,
max(close) AS highest_close,
min(close) AS lowest_close,
round(avg(daily_return_pct), 2) AS avg_return,
sum(volume) AS total_volume
FROM stock_market
GROUP BY ticker, company, sector;
'''
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:1234@localhost:3306/stock_analysis"
)

query = """
select *
from stock_performance
order by avg_return DESC
"""

df = pd.read_sql(query, engine)

print(df.head())
```

```python
from matplotlib.ticker import FuncFormatter

# df = DataFrame containing your stock_performance SQL view
df = df.copy()

# Clean data
df = df.dropna(subset=[
    "avg_close", "avg_return", "total_volume", "sector"
])

# Professional dark theme
sns.set_theme(
    style="dark",
    context="talk",
    rc={
        "figure.facecolor": "#0B1220",
        "axes.facecolor": "#111C30",
        "axes.edgecolor": "#506078",
        "axes.labelcolor": "white",
        "text.color": "white",
        "xtick.color": "#CBD5E1",
        "ytick.color": "#CBD5E1",
        "grid.color": "#334155"
    }
)

fig, ax = plt.subplots(figsize=(17, 9))

sns.scatterplot(
    data=df,
    x="avg_close",
    y="avg_return",
    hue="sector",
    size="total_volume",
    sizes=(100, 900),
    palette="bright",
    alpha=0.8,
    edgecolor="white",
    linewidth=0.8,
    ax=ax
)

# Label only top 10 stocks to avoid clutter
top_stocks = df.nlargest(10, "avg_return")

for _, row in top_stocks.iterrows():
    ax.annotate(
        row["ticker"].replace(".NS", ""),
        (row["avg_close"], row["avg_return"]),
        xytext=(7, 7),
        textcoords="offset points",
        fontsize=9,
        color="white",
        fontweight="bold"
    )

# Zero return reference line
ax.axhline(
    0,
    color="#FB7185",
    linestyle="--",
    linewidth=1.5,
    alpha=0.8
)

ax.set_title(
    "NIFTY 50 | STOCK PERFORMANCE OVERVIEW",
    fontsize=21,
    fontweight="bold",
    color="white",
    pad=25
)

ax.set_xlabel("Average Closing Price (₹)", fontsize=13)
ax.set_ylabel("Average Daily Return (%)", fontsize=13)

ax.xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"₹{x:,.0f}")
)

ax.grid(True, linestyle="--", alpha=0.25)
ax.set_axisbelow(True)

# Legend outside chart
legend = ax.legend(
    title="Sector / Total Volume",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    fontsize=9,
    title_fontsize=10,
    frameon=True
)

legend.get_frame().set_facecolor("#1E293B")
legend.get_frame().set_edgecolor("#475569")

for text in legend.get_texts():
    text.set_color("white")

legend.get_title().set_color("white")

plt.tight_layout()

plt.savefig(
    "nifty50_performance.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.show()
```
